```python
# Import Modular AI Engine
import modular.engine as me
```

```python
# Loading pytorch model is similar. Only addition is a input spec parameter at load
session = me.InferenceSession()
torch_options = me.TorchLoadOptions()
torch_options.input_specs = [me.TensorSpec(shape=[1, 16], dtype=me.DType.int32, name='attention_mask'), me.TensorSpec(shape=[1, 3, 224, 224], dtype=me.DType.float32, name='pixel_values'), me.TensorSpec(shape=[1, 16], dtype=me.DType.int32, name='input_ids')]
clip_torch = session.load('clip-vit.torchscript', torch_options)
```

```python
# The same APIs work for PyTorch also
import numpy as np
pixel_values = np.random.rand(1, 3, 224, 224).astype(np.float32)
attention_mask = np.ones((1, 16), dtype=np.int32)
input_ids = np.ones((1, 16), dtype=np.int32)
outputs_torch = clip_torch.execute(mask=attention_mask, pixel_values=pixel_values, input_ids=input_ids)
```
