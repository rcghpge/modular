```python
import subprocess
import tarfile
```

```python
def download_clip_vit_models():
    # Download pytorch model
    subprocess.run(['aws', 's3', 'cp', 's3://modular-model-storage/TorchScript/clip-vit-large-patch14-flattened.torchscript', 'clip-vit.torchscript'])

    # Download and extract tensorflow model
    subprocess.run(['aws', 's3', 'cp', 's3://modular-model-storage/Tensorflow/clip_vit_large_patch14_savedmodel.tar.gz', 'clip-vit-tf.tar.gz'])

    with tarfile.open('clip-vit-tf.tar.gz') as tar:
        tar.extractall('clip-vit-tf')
```

```python
download_clip_vit_models()
```
